Q2:- Using an active mysq12 connection, write a function to fetch all rows from a products table and log the results.
* Ans:- Here is a Node.js example using an active mysql2 connection to fetch all rows from the products table and log the results.

```python
const mysql = require('mysql2');

// Create connection
const connection = mysql.createConnection({
    host: 'localhost',
    user: 'root',
    password: 'your_password',
    database: 'app_db'
});

// Function to fetch all products
function getAllProducts() {
    const sql = 'SELECT * FROM products';

    connection.query(sql, (err, results) => {
        if (err) {
            console.error('Error fetching products:', err.message);
            return;
        }

        console.log('All Products:', results);
    });
}

// Establish connection and fetch products
connection.connect((err) => {
    if (err) {
        console.error('Connection failed:', err.message);
        return;
    }

    console.log('Connected to database');

    getAllProducts();
});

// run the File--
```

Q3:-Define a Sequelize model for a Task that has a title (string) and a completed (boolean) field.
* Ans:-here is how to define a task model using sequelize in node.js
* step1:-install sequelize- npm install sequelize mysql2
* step2:-Define the Task model-
Create a file named Task.js:

```python
const { DataTypes } = require('sequelize');
const sequelize = require('./database');

const Task = sequelize.define('Task', {
    title: {
        type: DataTypes.STRING,
        allowNull: false
    },
    completed: {
        type: DataTypes.BOOLEAN,
        defaultValue: false
    }
});

module.exports = Task;
```

* step3:- database connection (database.js)

```python
const { Sequelize } = require('sequelize');

const sequelize = new Sequelize('app_db', 'root', 'your_password', {
    host: 'localhost',
    dialect: 'mysql'
});

module.exports = sequelize;
```

* step4:-sync the model-

```python
const sequelize = require('./database');
const Task = require('./Task');

sequelize.sync()
    .then(() => {
        console.log('Task table created successfully');
    })
    .catch(err => {
        console.error('Error:', err);
    });
```

Q4:- Using a Sequelize model named Task, write a query to find all tasks and log their titles.
* Ans:-Using the Sequelize model Task, you can use the findAll() method to fetch all tasks and log their titles.
* Explanation:
Task.findAll() retrieves all rows from the Tasks table.
forEach() iterates through the results.
task.title logs each task's title.

```python
const Task = require('./Task');

async function getAllTasks() {
    try {
        const tasks = await Task.findAll();

        tasks.forEach(task => {
            console.log(task.title);
        });

    } catch (error) {
        console.error('Error fetching tasks:', error);
    }
}

getAllTasks();
```

Q7:- Using the native MongoDB driver, insert a new document into a users collection with a name of "John Doe" and an age of 30.
* Ans:-Insert a document using the native MongoDB driver
* Explanation:
db.collection('users') selects the users collection.
insertOne() inserts a single document.
result.insertedId logs the inserted document's ID.

```python
const { MongoClient } = require('mongodb');

async function insertUser() {
    const client = new MongoClient('mongodb://localhost:27017');

    try {
        await client.connect();

        const db = client.db('inventory');
        const users = db.collection('users');

        const result = await users.insertOne({
            name: "John Doe",
            age: 30
        });

        console.log("User inserted:", result.insertedId);

    } catch (error) {
        console.error(error);
    } finally {
        await client.close();
    }
}

insertUser();
```

Q9:- Write a MongoDB query to update a user's age to 31, finding the user by their name, "John Doe".
* Ans:-Explanation:
updateOne() updates the first matching document.
$set changes the user's age to 31.

```python
db.users.updateOne(
    { name: "John Doe" },
    { $set: { age: 31 } }
);
```
